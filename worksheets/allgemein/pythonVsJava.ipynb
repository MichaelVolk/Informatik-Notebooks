## 3. Methoden

In Java braucht jede Methode einen **Rückgabetyp** und **typisierte Parameter**. In Python reicht das Schlüsselwort `def` - Typen werden nicht angegeben. Methoden in Python können dadurch vielseitiger eingesetzt werden, der Methodenkopf ist dafür aber nicht mehr ganz so aussagekräftig.

**Java:**
```java
int verdopple(int zahl) {
    return zahl * 2;
}
```

**Python:**
```python
def verdopple(zahl):
    return zahl * 2
```

Der **Doppelpunkt** `:` ersetzt die öffnende geschweifte Klammer. Die **Einrückung** (4 Leerzeichen oder Tab) übernimmt die Rolle der geschweiften Klammern `{}`.

## 4. for-Schleifen

In Java besteht eine for-Schleife aus **Startwert**, **Bedingung** und **Schritt**. Python verwendet die Funktion `range()`, die diese drei Teile zusammenfasst.

**Java:**
```java
for (int i = 0; i < 5; i++) {
    println(i);
}
```

**Python:**
```python
for i in range(5):
    print(i)
```

`range(5)` erzeugt die Zahlen 0, 1, 2, 3, 4. Für einen anderen Startpunkt kann man `range(start, stop)` verwenden - `range(1, 6)` zählt zum Beispiel von 1 bis 5.

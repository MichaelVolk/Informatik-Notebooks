Die rote Fehlermeldung (auch **Traceback** genannt) enthält drei wichtige Informationen:

1. **Wo** ist der Fehler? → Der Pfeil (`--->`) zeigt die Zeile mit dem Problem.
2. **Was** für ein Fehler? → Am Ende steht der Fehlertyp, z.B. `NameError`.
3. **Warum** ein Fehler? → Danach folgt eine Erklärung, z.B. `name 'summe' is not defined`.

Im Beispiel oben sagt Python: *"Ich kenne keine Variable namens 'summe'."* Der Grund ist, dass `summe` nie einen Wert bekommen hat – `x` und `y` existieren, aber `summe` nicht. Ergänze nun in der obigen Box eine variable `summe`, in der die Summe von x und y berechnet wird.

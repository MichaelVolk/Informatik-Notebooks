## Beispiel: Bevölkerung und CO₂-Emissionen

Bevor du selbst loslegst, ein Beispiel. Wir schauen uns den Zusammenhang zwischen Bevölkerung und CO₂-Emissionen an – intuitiv sollte mehr Bevölkerung auch mehr Emissionen bedeuten.

<div class="alert alert-block alert-warning">
    Führe den folgenden Code-Block aus und schau dir das Ergebnis an.
</div>

````python
x, y = filtere(daten["bevoelkerung"], daten["co2"])

plt.scatter(x, y)
plt.title("Bevölkerung vs. CO₂-Emissionen")
plt.xlabel("Bevölkerung")
plt.ylabel("CO₂-Emissionen (kt)")
plt.show()
````

Fast alle Punkte kleben in der linken unteren Ecke – nur wenige extreme Ausreißer (China, USA, Indien...) sind weit weg. Der Zusammenhang ist kaum zu erkennen.

Das Problem: Die Werte liegen in völlig unterschiedlichen **Größenordnungen**. Die Bevölkerungen reichen von einigen Hunderttausend bis über 1,4 Milliarden – ein Faktor von mehr als 10.000. Auf einer normalen linearen Achse bekommt jede Zahl gleich viel Platz. Dadurch werden kleine Länder in eine unsichtbare Ecke gequetscht.

Die Lösung: eine **logarithmische Skala**. Auf ihr bekommt jede Größenordnung gleich viel Platz – der Abstand von 1.000 zu 10.000 ist genauso groß wie von 10.000 zu 100.000. In Python schaltet man sie mit zwei Zeilen um:

```python
plt.xscale("log")
plt.yscale("log")
```

<div class="alert alert-block alert-warning">
    Führe den folgenden Code-Block aus. Was verändert sich im Vergleich zum Diagramm oben?
</div>

````python
x, y = filtere(daten["bevoelkerung"], daten["co2"])

plt.scatter(x, y)
plt.xscale("log")
plt.yscale("log")
plt.title("Bevölkerung vs. CO₂-Emissionen (logarithmische Skala)")
plt.xlabel("Bevölkerung (log)")
plt.ylabel("CO₂-Emissionen in kt (log)")
plt.show()
````

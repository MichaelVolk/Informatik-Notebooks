## 1. Was sind Libraries?

Python selbst kann schon viel, z.B. Berechnungen anstellen, Text ausgeben, Schleifen laufen lassen. Für spezialisierte Aufgaben wie das Erstellen von Diagrammen wäre es aber sehr mühsam, alles von Grund auf selbst zu programmieren.

Genau hier kommen **Libraries** (dt. *Bibliotheken*) ins Spiel:

> Eine Library ist eine Sammlung von fertig geschriebenem Code, den andere Programmierer:innen entwickelt haben und den du in deinem eigenen Programm nutzen kannst.

**Beispiel:** Wenn du ein Videospiel spielst, wurden tausende Stunden in Engine, Physik-Simulation und Grafik investiert. Als Spieleentwickler:in nutzt du diese fertigen Systeme und baust "nur" das eigentliche Spiel darauf. Libraries funktionieren genauso.

In Python gibt es Libraries für fast jeden Zweck:

| Library | Zweck |
|---------|-------|
| `matplotlib` | Diagramme und Grafiken erstellen |
| `pandas` | Große Datensätze verwalten und auswerten |
| `numpy` | Mathematische Berechnungen mit großen Zahlenmengen |
| `requests` | Daten aus dem Internet abrufen |

### Library importieren

Bevor du eine Library nutzen kannst, musst du sie **importieren**. Das funktioniert mit dem Schlüsselwort `import`:

```python
import matplotlib.pyplot as plt
```

Das `as plt` legt eine **Abkürzung** fest: Statt später immer `matplotlib.pyplot.plot(...)` schreiben zu müssen, reicht danach `plt.plot(...)`. Die Abkürzung `plt` für matplotlib ist dabei so weit verbreitet, dass sie als Standard gilt.

    Führe den folgenden Code-Block aus, um matplotlib zu importieren. Das erlaubt es dir, in allen weiteren Blöcken die Funktionen der Library matplotlib zu nutzen.

## 2. Balkendiagramm

Ein **Balkendiagramm** eignet sich gut, um einzelne Werte miteinander zu vergleichen. Zum Beispiel: Welche Apps nutzen Jugendliche am häufigsten?

Der Befehl `plt.bar()` erstellt ein Balkendiagramm:

```python
plt.bar(kategorien, werte)
```

Mit weiteren Befehlen lässt sich das Diagramm beschriften:

| Befehl | Bedeutung |
|--------|-----------|
| `plt.title("...")` | Titel des Diagramms |
| `plt.xlabel("...")` | Beschriftung der x-Achse |
| `plt.ylabel("...")` | Beschriftung der y-Achse |
| `plt.show()` | Diagramm anzeigen |

<div class="alert alert-block alert-warning">
    Führe den folgenden Code-Block aus und schau dir das Ergebnis an.
</div>

````python
# Daten: Anteil der 12-19-Jährigen in Deutschland, die diese App täglich nutzen (in %)
apps = ["WhatsApp", "YouTube", "Instagram", "TikTok", "Snapchat"]
nutzung = [93, 88, 72, 67, 54]

plt.bar(apps, nutzung)
plt.title("Tägliche App-Nutzung Jugendlicher in Deutschland (2024)")
plt.ylabel("Nutzeranteil in %")
plt.show()
````

## 3. Liniendiagramm

Ein **Liniendiagramm** zeigt, wie sich ein Wert über die Zeit verändert. Es eignet sich besonders gut für Trends. Zum Beispiel: Wie hat sich die tägliche Bildschirmzeit von Jugendlichen in den letzten Jahren entwickelt?

Der Befehl `plt.plot()` erstellt ein Liniendiagramm:

```python
plt.plot(x_werte, y_werte)
```

Mit `marker="o"` werden die einzelnen Messpunkte als Kreise markiert:

```python
plt.plot(x_werte, y_werte, marker="o")
```

<div class="alert alert-block alert-warning">
    Führe den folgenden Code-Block aus und schau dir das Ergebnis an.
</div>

````python
# Durchschnittliche tägliche Bildschirmzeit von Jugendlichen (12-19 Jahre) in Deutschland
jahre = [2019, 2020, 2021, 2022, 2023, 2024]
stunden = [3.5, 4.2, 4.8, 5.1, 5.4, 5.8]

plt.plot(jahre, stunden, marker="o")
plt.title("Tägliche Bildschirmzeit Jugendlicher in Deutschland")
plt.xlabel("Jahr")
plt.ylabel("Stunden pro Tag")
plt.show()
````

## 4. Kreisdiagramm

Ein **Kreisdiagramm** (auch *Tortendiagramm*) zeigt, wie sich ein Ganzes in Teile aufteilt. Zum Beispiel: Welche Musikgenres hören Jugendliche am liebsten?

Der Befehl `plt.pie()` erstellt ein Kreisdiagramm:

```python
plt.pie(werte, labels=beschriftungen)
```

Mit dem Parameter `autopct` werden die Prozentzahlen direkt im Diagramm angezeigt:

```python
plt.pie(werte, labels=beschriftungen, autopct="%1.1f%%")
```

`"%1.1f%%"` ist ein Formatierungsbefehl: Er zeigt eine Nachkommastelle an und hängt ein %-Zeichen an.

<div class="alert alert-block alert-warning">
    Führe den folgenden Code-Block aus und schau dir das Ergebnis an.
</div>

````python
# Umfrage: Welche Musikgenres hören Jugendliche am liebsten? (in %)
genres = ["Pop", "Hip-Hop / Rap", "Electronic", "Rock", "Sonstige"]
anteile = [34, 28, 15, 13, 10]

plt.pie(anteile, labels=genres, autopct="%1.1f%%")
plt.title("Lieblingsmusikgenres Jugendlicher (2024)")
plt.show()
````

## 5. Scatter Plot

Ein **Scatter Plot** (Streudiagramm) eignet sich, wenn du den Zusammenhang zwischen zwei Variablen untersuchen willst – zum Beispiel: Lässt die Akkukapazität mit dem Alter eines Handys nach?

Anders als beim Balken- oder Liniendiagramm hat jeder Punkt **zwei Koordinaten**: einen Wert auf der x-Achse und einen auf der y-Achse. Die Form der Punktwolke verrät, ob und wie die beiden Variablen zusammenhängen.

Der Befehl `plt.scatter()` erstellt einen Scatter Plot:

```python
plt.scatter(x_werte, y_werte)
```

<div class="alert alert-block alert-warning">
    Führe den folgenden Code-Block aus und schau dir das Ergebnis an.
</div>

````python
# Alter eines Handys (in Monaten) vs. verbleibende Akkukapazität (in %)
handyalter = [3, 6, 9, 12, 15, 18, 21, 24, 30, 36]
akkukapazitaet = [98, 95, 91, 87, 83, 78, 73, 69, 60, 52]

plt.scatter(handyalter, akkukapazitaet)
plt.title("Handy-Alter und Akkukapazität")
plt.xlabel("Alter des Handys (Monate)")
plt.ylabel("Akkukapazität (%)")
plt.show()
````
